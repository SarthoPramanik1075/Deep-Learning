```python
def text_to_indices(sentence, vocab):
  numerical_sentences=[]
  for token in sentence:
    # print(token)
    if token in vocab:
      numerical_sentences.append(vocab[token])
    else:
      numerical_sentences.append(vocab['<unk>'])
  return numerical_sentences
```

```python
input_numerical_sentence=[]
for sentence in input_sentences:
  # print(sentence)
  input_numerical_sentence.append(text_to_indices(word_tokenize(sentence.lower()),vocab))
  # print(word_tokenize(sentence.lower()))
```

#Custom Dataset & DataLoader

```python
class CustomDataset(Dataset):
  def __init__(self, X, y):
    self.X = X
    self.y = y

  def __len__(self):
    return len(self.X)

  def __getitem__(self, idx):
    return self.X[idx], self.y[idx]
```

```python
dataset = CustomDataset(X,y)
```

```python
len(dataset)
```

```python
dataloader = DataLoader(dataset, batch_size=32, shuffle=True)
```

##LSTM Cell

```python
class LSTMModel(nn.Module):
  def __init__(self,vocab_size):
    super().__init__()
    self.embedding = nn.Embedding(vocab_size,100)
    self.lstm = nn.LSTM(100,150,batch_first=True)
    self.fc = nn.Linear(150,vocab_size)

  def forward(self,x):
    embedded = self.embedding(x)
    intermediate_hidden_states,(final_hidden_state,final_cell_state) = self.lstm(embedded)
    output = self.fc(final_hidden_state.squeeze(0))
    return output
```

#Training Loop

```python
epochs = 50
learning_rate=0.001
criterion = nn.CrossEntropyLoss()
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr = learning_rate)
```

```python
for epoch in range(epochs):
  total_loss=0

  for batch_x, batch_y in dataloader:
    batch_x = batch_x.to(device)
    batch_y = batch_y.to(device)
    optimizer.zero_grad()
    output = model(batch_x)

    loss = criterion(output, batch_y)
    loss.backward()
    optimizer.step()
    total_loss += loss.item()
    print(f"Epoch {epoch+1}, Loss={total_loss:.4f}")
```

```python
def prediction(model, vocab, text):
  #tokenize
  tokenized_text = word_tokenize(text.lower())

  #print(tokenized_text)

  #numerical indies
  numerical_text = text_to_indices(tokenized_text, vocab)
  # print(numerical_text)

  #padding
  padded_text = torch.tensor([0]* (25-len(numerical_text)) + numerical_text, dtype= torch.long).unsqueeze(0)
  # print(padded_text)

  output = model(padded_text)
  # print("model's output", output)


  #predicted index

  value, index = torch.max(output,dim=1)

  # print(f"printing index {index}")
  # print("printing the word")

  return text + " " + list(vocab.keys())[index]
```

#Accuracy Calculation


```python
def calculate_accuracy(model, vocab, text):
  model.eval()

  correct = 0
  total = 0

  with torch.no_grad():
    for batch_x, batch_y in dataloader:
      batch_x = batch_x.to(device)
      batch_y = batch_y.to(device)

      output = model(batch_x)

      _, predicted = torch.max(output, dim=1)
      # print("printing predicted", predicted)

      correct += (predicted == batch_y).sum().item()
      # print(correct)

      total += batch_y.size(0)
      # print(total)

  accuracy = correct / total
  return accuracy
```

```python
accuracy= calculate_accuracy(model, dataloader, device)
print(f"Accuracy: {accuracy:.2f}%")
```
